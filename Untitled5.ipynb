```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# ----------------------
# Step 1: Enter the data
# ----------------------
data = []

# Blend X
blend_x = [
    [123, 128, 166, 151],
    [156, 150, 178, 125],
    [112, 174, 187, 117],
    [100, 116, 153, 155],
    [168, 109, 195, 158]
]

# Blend Y
blend_y = [
    [135, 175, 140, 167],
    [130, 132, 145, 183],
    [176, 120, 159, 142],
    [120, 187, 131, 167],
    [155, 184, 126, 168]
]

# Blend Z
blend_z = [
    [156, 186, 185, 175],
    [180, 138, 206, 173],
    [147, 178, 188, 154],
    [146, 176, 165, 191],
    [193, 190, 188, 169]
]

# Combine into a list of dictionaries
fertilizers = ['X', 'X', 'X', 'X', 'X',
               'Y', 'Y', 'Y', 'Y', 'Y',
               'Z', 'Z', 'Z', 'Z', 'Z']
crops = ['Wheat', 'Corn', 'Soy', 'Rice']

for fert, rows in zip(['X']*5 + ['Y']*5 + ['Z']*5, blend_x + blend_y + blend_z):
    for crop, yield_val in zip(crops, rows):
        data.append({'Fertilizer': fert, 'Crop': crop, 'Yield': yield_val})

df = pd.DataFrame(data)

# ----------------------
# Step 2: Run Two-Way ANOVA
# ----------------------
model = ols('Yield ~ C(Fertilizer) + C(Crop) + C(Fertilizer):C(Crop)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print("ANOVA Results:")
print(anova_table.round(4))
print("\n")

# ----------------------
# Step 3: Descriptive Statistics
# ----------------------
desc_stats = df.groupby(['Fertilizer', 'Crop'])['Yield'].agg(['mean', 'std']).round(2)
print("Descriptive Statistics (Mean & Std Dev):")
print(desc_stats)
print("\n")

# ----------------------
# Step 4: Post-hoc Tests (if significant)
# ----------------------
if anova_table.loc['C(Fertilizer)', 'PR(>F)'] < 0.05:
    print("Post-hoc test for Fertilizer:")
    tukey_fert = pairwise_tukeyhsd(endog=df['Yield'], groups=df['Fertilizer'], alpha=0.05)
    print(tukey_fert)
    print("\n")

if anova_table.loc['C(Crop)', 'PR(>F)'] < 0.05:
    print("Post-hoc test for Crop:")
    tukey_crop = pairwise_tukeyhsd(endog=df['Yield'], groups=df['Crop'], alpha=0.05)
    print(tukey_crop)
    print("\n")

if anova_table.loc['C(Fertilizer):C(Crop)', 'PR(>F)'] < 0.05:
    print("Interaction is significant; simple effects analysis recommended.")
```
